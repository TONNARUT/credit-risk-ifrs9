```python
import pandas as pd
from sklearn.model_selection import train_test_split
df = pd.read_csv("../data/application_train.csv")

# 1. 365243 in days employed is considered anomaly
df["EMPLOYED_ANOM"] = (df["DAYS_EMPLOYED"] == 365243).astype(int)
df.loc[df["EMPLOYED_ANOM"] == 1, "DAYS_EMPLOYED"] = None # Data transformation to indicate variable, if its 1 then its anomaly 

# 2. Readable features
df["AGE_YEARS"] = -df["DAYS_BIRTH"] / 365
df["YEARS_EMPLOYED"] = -df["DAYS_EMPLOYED"] / 365
df["CREDIT_INCOME_RATIO"] = df["AMT_CREDIT"] / df["AMT_INCOME_TOTAL"]
df["ANNUITY_INCOME_RATIO"] = df["AMT_ANNUITY"] / df["AMT_INCOME_TOTAL"]

# 3. Drop duplicate building stats (keep _AVG) and 4 rows with gender 'XNA'
df = df.drop(columns=[c for c in df.columns if c.endswith(("_MODE", "_MEDI"))])
df = df[df["CODE_GENDER"] != "XNA"]

# 4. Split 70/30, keeping the 8% default rate in both
train, test = train_test_split(df, test_size=0.3, stratify=df["TARGET"], random_state=42) # stratify will tell that train and test have the same target mean
train.to_csv("../data/train_clean.csv", index=False)
test.to_csv("../data/test_clean.csv", index=False)
print(train.shape, test.shape, train["TARGET"].mean(), test["TARGET"].mean())
```
